# Mathematics and Statistics for Data Science

This notebook follows a teaching pattern: start with a single metric, then add a second concept, then close with a short decision task grounded in a real dataset.


## Mean, median, and spread

We start with one dataset and compute several summaries. Each metric answers a slightly different question.


```python
import numpy as np

wait_times = np.array([4, 5, 5, 6, 10, 30], dtype=float)
print(np.mean(wait_times))
print(np.median(wait_times))
print(np.std(wait_times, ddof=1))
assert np.median(wait_times) == 5.5
assert np.mean(wait_times) > np.median(wait_times)
```

## Vector operations and linear combinations

Linear algebra is not only theory. Every time we train a model, we are combining features by dot products and weights.


```python
x = np.array([1.0, 2.0, 3.0])
w = np.array([0.5, 1.5, 2.0])
print(float(x @ w))
assert abs(float(x @ w) - 9.5) < 1e-9
```

## Probability and the idea of uncertainty

Probability is the language for uncertainty. Students usually understand it easier when they see a small, finite sample space.


```python
outcomes = ["red", "blue", "green", "red", "blue"]
prob_red = outcomes.count("red") / len(outcomes)
prob_blue_or_green = sum(v in ("blue", "green") for v in outcomes) / len(outcomes)
print(prob_red, prob_blue_or_green)
assert prob_red == 0.4
assert prob_blue_or_green == 0.6
```

## Sampling and the normal approximation

The normal distribution appears because sums and means of many independent influences tend to settle around a center. Use this section to emphasize that sample variation is a fact, not a mistake.


```python
rng = np.random.RandomState(42)
base = rng.normal(50, 10, size=5000)
means = [np.mean(rng.choice(base, size=25)) for _ in range(200)]
print(round(np.std(means), 3))
assert np.std(means) > 0
```

## Optimization and gradient descent

This is the moment when students see why mathematics matters in machine learning. A loss function tells us how wrong the current parameters are, and gradient descent moves them toward a better answer.


```python
def loss(x):
    return (x - 3.0) ** 2

def derivative(x):
    step = 1e-5
    return (loss(x + step) - loss(x - step)) / (2 * step)

value = -2.0
for _ in range(200):
    value -= 0.05 * derivative(value)
print(round(value, 4), round(loss(value), 8))
assert abs(value - 3.0) < 0.01
```

## Practice project: compare two service teams

This is a capstone-style short project: compare the two groups, explain the difference between mean and median, and discuss whether the observed difference is likely to be meaningful or simply a sample artifact.


```python
team_a = np.array([8, 10, 11, 12, 14, 16, 35], dtype=float)
team_b = np.array([7, 8, 9, 10, 11, 12, 18], dtype=float)
print(np.mean(team_a), np.median(team_a))
print(np.mean(team_b), np.median(team_b))
assert np.mean(team_b) < np.mean(team_a)
```
