# Neural Networks with NumPy

This notebook uses a tiny NumPy-only network to teach the idea of a forward pass, a loss, and a backward update without requiring a heavyweight framework.


## Activation and a basic neuron

Students should understand the neuron as a weighted sum + bias + nonlinear activation before they are asked to implement an entire network.


```python
import numpy as np

inputs = np.array([0.5, 1.0])
weights = np.array([0.8, -0.4])
bias = 0.1
linear = inputs @ weights + bias
out = 1 / (1 + np.exp(-linear))
print(round(float(linear), 3), round(float(out), 3))
assert 0 < out < 1
```

## XOR is nonlinear

The classic XOR example is the easiest way to show why a single linear neuron is not enough. The point is conceptual clarity, not large-scale training.


```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
y = np.array([0, 1, 1, 0], dtype=float)
print(X)
print(y)
assert X.shape == (4, 2)
```

## Forward pass and loss

The forward pass is where inputs become predictions. The loss function tells us how far those predictions are from the target.


```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def forward(x, w1, b1, w2, b2):
    hidden = np.tanh(x @ w1 + b1)
    out = sigmoid(hidden @ w2 + b2)
    return hidden, out

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
W1 = np.random.RandomState(10).normal(0, 0.5, size=(2, 2))
b1 = np.zeros((1, 2))
W2 = np.random.RandomState(11).normal(0, 0.5, size=(2, 1))
b2 = np.zeros((1, 1))
h, p = forward(X, W1, b1, W2, b2)
print(np.round(p.ravel(), 3))
assert p.shape == (4, 1)
```

## One brief training loop

The training loop is intentionally short. The goal is not to teach production deep learning here, but to show the mechanics of parameter updates.


```python
learning_rate = 0.3
for _ in range(200):
    hidden, pred = forward(X, W1, b1, W2, b2)
    grad = (pred - y.reshape(-1, 1)) / len(X)
    W2 -= learning_rate * (hidden.T @ grad)
    b2 -= learning_rate * grad.sum(axis=0, keepdims=True)
    hidden_grad = (grad @ W2.T) * (1 - hidden ** 2)
    W1 -= learning_rate * (X.T @ hidden_grad)
    b1 -= learning_rate * hidden_grad.sum(axis=0, keepdims=True)
print(np.round(forward(X, W1, b1, W2, b2)[1].ravel(), 3))
assert np.isfinite(np.round(forward(X, W1, b1, W2, b2)[1], 3)).all()
```
