# Python Foundations for Data Science

This notebook is intentionally built in a scaffolded way: a short code example first, then a slightly larger example, then a full mini-project. The objective is to teach the idea before asking students to combine many concepts at once.


## A small but realistic workflow

We first work with a tiny list of order records. The data is simple enough to inspect manually, but still realistic enough to represent the kind of records a data analyst receives from a business system.


```python
orders = [
    {"item": "Notebook", "quantity": 3, "unit_price": 2.5},
    {"item": "Pen", "quantity": 5, "unit_price": 1.25},
    {"item": "Coffee", "quantity": 10, "unit_price": 3.4},
]
print(orders[0])
print(orders[0]["quantity"] * orders[0]["unit_price"])
assert round(orders[0]["quantity"] * orders[0]["unit_price"], 2) == 7.5
```

## Working with types and conversion

When data comes from a CSV or a form, numbers often arrive as strings. A common source of bugs is forgetting to convert strings before computing with them.


```python
raw_quantity = "7"
raw_price = "1.50"
quantity = int(raw_quantity)
price = float(raw_price)
print(quantity * price)
assert quantity * price == 10.5
```

## Decisions and repeats

A simple if/else statement decides which branch to take. A for loop is the standard way to inspect every item in a collection. This is the first step toward more useful automation.


```python
totals = []
for record in orders:
    subtotal = record["quantity"] * record["unit_price"]
    if subtotal >= 10:
        totals.append("high")
    else:
        totals.append("low")
print(totals)
assert totals == ["low", "low", "high"]
```

## Lists, dictionaries, and functions

At this stage, students should begin to see that functions are not decoration. They are the mechanism that turns raw data into a clean reusable result.


```python
def subtotal(record):
    return record["quantity"] * record["unit_price"]

def total_revenue(records):
    return sum(subtotal(record) for record in records)

print(total_revenue(orders))
assert round(total_revenue(orders), 2) == 47.75
```

## Validation and a realistic project

The next block introduces a validation layer. A good student project does not just print data; it verifies assumptions and raises errors when inputs are no longer valid.


```python
def build_report(records, tax_rate=0.0):
    if tax_rate < 0:
        raise ValueError("tax rate cannot be negative")
    subtotal = 0.0
    by_item = {}
    for record in records:
        if "item" not in record or "quantity" not in record or "unit_price" not in record:
            raise ValueError("missing fields in a record")
        item = record["item"]
        amount = record["quantity"] * record["unit_price"]
        subtotal += amount
        by_item[item] = by_item.get(item, 0.0) + amount
    total = subtotal * (1 + tax_rate)
    return {
        "subtotal": round(subtotal, 2),
        "tax": round(subtotal * tax_rate, 2),
        "total": round(total, 2),
        "by_item": {key: round(value, 2) for key, value in by_item.items()},
    }

report = build_report(orders, tax_rate=0.08)
print(report)
assert report["subtotal"] == 47.75
assert report["total"] == 51.57
```
